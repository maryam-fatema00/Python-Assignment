```python
# Q15

def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even_odd(14))
print(check_even_odd(7))
```

```python
# Q18

def calculate_total_sum(limit):
    total = 0

    for i in range(limit):
        total += i

    return total

result = calculate_total_sum(10)

print(result)
```

```python
# Q19

def grade_evaluator(name, marks):
    if marks >= 50:
        return f"{name} passed with {marks} marks."
    else:
        return f"{name} failed with {marks} marks."

print(grade_evaluator("Ali", 75))
print(grade_evaluator("Ahmed", 42))
```

```python
# Q20

def repeat_greeting(name, count):
    if count <= 0:
        return "Invalid count."
    else:
        for i in range(count):
            print(f"Hello, {name}!")

repeat_greeting("Usman", 3)
```
